```python
# ==========================================================
# Experiment: Morphological Analysis using Python (NLTK)
# Aim:
# Analyze words by identifying prefixes, stems, suffixes,
# stemming, and lemmatization.
# ==========================================================

# Install NLTK (Run only once in Colab)
# !pip install nltk

import nltk
from nltk.stem import PorterStemmer, SnowballStemmer, WordNetLemmatizer

# Download required resources (Run once)
nltk.download('wordnet')
nltk.download('omw-1.4')

print("Rishabh Raikwar 23BAI70136")

# ----------------------------------------------------------
# Sample Dataset
# ----------------------------------------------------------
words = [
    "unhappy",
    "replayed",
    "kindness",
    "studies",
    "running",
    "nationalization",
    "children",
    "playing",
    "redo",
    "disagree",
    "careless",
    "beautifully"
]

# ----------------------------------------------------------
# Prefixes and Suffixes (Simple Rule-Based)
# ----------------------------------------------------------
prefixes = [
    "un", "re", "dis", "pre", "mis", "non",
    "in", "im", "ir", "il", "over", "under"
]

suffixes = [
    "ing", "ed", "ly", "ness", "tion",
    "s", "es", "ment", "able", "ful",
    "less", "ize", "ization"
]

# ----------------------------------------------------------
# Initialize Stemmers and Lemmatizer
# ----------------------------------------------------------
porter = PorterStemmer()
snowball = SnowballStemmer("english")
lemmatizer = WordNetLemmatizer()

# ----------------------------------------------------------
# Function for Morphological Analysis
# ----------------------------------------------------------
def analyze_word(word):
    prefix = "-"
    suffix = "-"
    stem = word

    # Detect Prefix
    for p in prefixes:
        if word.startswith(p) and len(word) > len(p):
            prefix = p
            stem = word[len(p):]
            break

    # Detect Suffix
    for s in sorted(suffixes, key=len, reverse=True):
        if stem.endswith(s) and len(stem) > len(s):
            suffix = s
            stem = stem[:-len(s)]
            break

    porter_stem = porter.stem(word)
    snowball_stem = snowball.stem(word)

    # Lemmatization
    lemma = lemmatizer.lemmatize(word)

    return [
        word,
        prefix,
        stem,
        suffix,
        porter_stem,
        snowball_stem,
        lemma
    ]

# ----------------------------------------------------------
# Display Results
# ----------------------------------------------------------
print("=" * 110)
print("{:<18}{:<12}{:<18}{:<12}{:<18}{:<18}{:<18}".format(
    "Word", "Prefix", "Stem", "Suffix",
    "Porter Stem", "Snowball Stem", "Lemma"))
print("=" * 110)

for w in words:
    result = analyze_word(w)
    print("{:<18}{:<12}{:<18}{:<12}{:<18}{:<18}{:<18}".format(*result))

print("=" * 110)
```

```python
# ============================================
# BIGRAM LANGUAGE MODEL
# Zero-Frequency Problem & Add-1 Smoothing
# ============================================

from collections import Counter
print("Adarsh, 23bai70091")

# Step 1: Training data
text = "I love NLP I love Python I like AI"

# Convert to lowercase and split into words
words = text.lower().split()

print("Words:")
print(words)


# Step 2: Create bigrams
bigrams = []

for i in range(len(words) - 1):
    bigrams.append((words[i], words[i + 1]))

print("\nBigrams:")
print(bigrams)


# Step 3: Count bigrams
bigram_counts = Counter(bigrams)

print("\nBigram Counts:")
print(bigram_counts)


# Step 4: Count unigrams
unigram_counts = Counter(words)

print("\nUnigram Counts:")
print(unigram_counts)


# Step 5: Vocabulary size
vocabulary = set(words)
V = len(vocabulary)

print("\nVocabulary:")
print(vocabulary)

print("Vocabulary Size:", V)


# Step 6: Bigram probability without smoothing
def bigram_probability(w1, w2):
    return bigram_counts[(w1, w2)] / unigram_counts[w1]


print("\nWithout Smoothing:")

print("P(love | i) =",
      bigram_probability("i", "love"))

print("P(ai | love) =",
      bigram_probability("love", "ai"))


# Step 7: Add-1 (Laplace) smoothing
def smoothed_bigram_probability(w1, w2):
    return (bigram_counts[(w1, w2)] + 1) / \
           (unigram_counts[w1] + V)


print("\nWith Add-1 Smoothing:")

print("P(love | i) =",
      smoothed_bigram_probability("i", "love"))

print("P(ai | love) =",
      smoothed_bigram_probability("love", "ai"))


# Step 8: Comparison
print("\nComparison:")

print("Without smoothing:",
      bigram_probability("love", "ai"))

print("With Add-1 smoothing:",
      smoothed_bigram_probability("love", "ai"))

# ============================================
# MLE vs Laplace vs Kneser-Ney
# ============================================

from nltk.lm import MLE, Laplace, KneserNeyInterpolated
from nltk.lm.preprocessing import padded_everygram_pipeline


# Training corpus
corpus = [
    "the cat sat on the mat".split(),
    "the dog sat on the rug".split(),
    "the cat chased the mouse".split(),
    "the dog chased the ball".split()
]

# Trigram model
N = 3


# Function to train model
def train(model):

    train_data, vocab = padded_everygram_pipeline(
        N, corpus
    )

    model.fit(train_data, vocab)

    return model


# Train three models
models = {
    "MLE": train(MLE(N)),
    "Laplace": train(Laplace(N)),
    "KneserNey": train(KneserNeyInterpolated(N))
}


# Test cases
tests = [
    ("cat", ["the"]),
    ("mouse", ["the", "chased"]),
    ("rug", ["the", "cat"])
]


# Display probabilities
for name, model in models.items():

    print("\n", name)

    for word, context in tests:

        probability = model.score(
            word,
            context
        )

        print(
            word,
            "|",
            context,
            "=",
            round(probability, 6)
        )
```

```python
# ============================================================
# HIDDEN MARKOV MODEL FOR POS TAGGING
# ============================================================

import nltk
from nltk.corpus import treebank
from nltk.tag.hmm import HiddenMarkovModelTrainer

print("Adarsh 23BAI70091")

# Download Treebank dataset
nltk.download("treebank", quiet=True)

# ------------------------------------------------------------
# 1. Load Dataset
# ------------------------------------------------------------

sentences = treebank.tagged_sents()[:2500]

# 80% Training, 20% Testing
split = int(0.8 * len(sentences))

train_sents = sentences[:split]
test_sents = sentences[split:]

print("Training sentences:", len(train_sents))
print("Testing sentences:", len(test_sents))


# ------------------------------------------------------------
# 2. Train HMM
# ------------------------------------------------------------

trainer = HiddenMarkovModelTrainer()

hmm_tagger = trainer.train_supervised(train_sents)

print("\nHMM training completed.")


# ------------------------------------------------------------
# 3. Display HMM States
# ------------------------------------------------------------

states = hmm_tagger._states

print("\nNumber of POS states:", len(states))

print("\nPOS States:")
print(states)


# ------------------------------------------------------------
# 4. Display Transition Probabilities
# ------------------------------------------------------------

print("\n========== TRANSITION PROBABILITIES ==========")

for tag in states[:10]:

    print("\nFrom:", tag)

    for next_tag in states[:10]:

        probability = hmm_tagger._transitions[tag].prob(next_tag)

        if probability > 0:
            print(
                tag,
                "->",
                next_tag,
                "=",
                round(probability, 4)
            )


# ------------------------------------------------------------
# 5. Display Emission Probabilities
# ------------------------------------------------------------

print("\n========== EMISSION PROBABILITIES ==========")

for tag in states[:10]:

    print("\nTag:", tag)

    count = 0

    # Use samples() instead of directly iterating
    for word in hmm_tagger._outputs[tag].samples():

        probability = hmm_tagger._outputs[tag].prob(word)

        if probability > 0:

            print(
                tag,
                "->",
                word,
                "=",
                round(probability, 4)
            )

            count += 1

        if count == 5:
            break


# ------------------------------------------------------------
# 6. Tag an Unseen Sentence
# ------------------------------------------------------------

sample = "The market opened higher today".split()

tagged_sentence = hmm_tagger.tag(sample)

print("\n========== POS TAGGING ==========")

print("Sentence:")
print(sample)

print("\nPredicted POS tags:")
print(tagged_sentence)


# ------------------------------------------------------------
# 7. VITERBI DECODING DEMONSTRATION
# ------------------------------------------------------------

print("\n========== VITERBI DECODING ==========")

states = ["N", "V"]

observations = ["time", "flies", "arrow"]

start = {
    "N": 0.6,
    "V": 0.4
}

transition = {
    "N": {"N": 0.3, "V": 0.7},
    "V": {"N": 0.8, "V": 0.2}
}

emission = {
    "N": {
        "time": 0.5,
        "flies": 0.4,
        "arrow": 0.6
    },

    "V": {
        "time": 0.1,
        "flies": 0.7,
        "arrow": 0.1
    }
}


# Viterbi tables
V = [{}]
path = {}


# Initialization
for state in states:

    V[0][state] = (
        start[state] *
        emission[state][observations[0]]
    )

    path[state] = [state]


# Recursion
for t in range(1, len(observations)):

    V.append({})

    new_path = {}

    for current_state in states:

        candidates = []

        for previous_state in states:

            probability = (
                V[t - 1][previous_state]
                * transition[previous_state][current_state]
                * emission[current_state][observations[t]]
            )

            candidates.append(
                (probability, previous_state)
            )

        best_probability, best_previous = max(candidates)

        V[t][current_state] = best_probability

        new_path[current_state] = (
            path[best_previous] + [current_state]
        )

    path = new_path


# Find best final state
best_probability, best_state = max(
    (V[-1][state], state)
    for state in states
)

print("Observations:", observations)

print(
    "Best tag sequence:",
    path[best_state]
)

print(
    "Best path probability:",
    round(best_probability, 8)
)


# ------------------------------------------------------------
# 8. Evaluation
# ------------------------------------------------------------

accuracy = hmm_tagger.accuracy(test_sents)

print("\n========== EVALUATION ==========")

print(
    "Test Accuracy:",
    round(accuracy, 4)
)
```

```python
import nltk
from nltk import word_tokenize, pos_tag, RegexpParser
from nltk.chunk import tree2conlltags
print("23bai70136 Rishabh Raikwar")

# Download required resources
for pkg in [
    "punkt",
    "punkt_tab",
    "averaged_perceptron_tagger",
    "averaged_perceptron_tagger_eng"
]:
    try:
        nltk.download(pkg, quiet=True)
    except Exception:
        pass


# Input sentence
sentence = "The intelligent student built a powerful language model"


# Tokenization
tokens = word_tokenize(sentence)


# POS Tagging
tagged = pos_tag(tokens)

print("Tagged Tokens:")
print(tagged)


# Chunk Grammar
chunk_grammar = r"NP: {<DT>?<JJ.*>*<NN.*>+}"


# Create chunk parser
chunker = RegexpParser(chunk_grammar)


# Perform chunking
tree = chunker.parse(tagged)

print("\nChunk Tree:")
print(tree)


# IOB representation
print("\nIOB Tags:")

for word, pos, iob in tree2conlltags(tree):
    print(f"{word:<14}{pos:<6}{iob}")


# Extract noun phrases
print("\nNoun Phrases:")

for subtree in tree.subtrees(
    filter=lambda t: t.label() == "NP"
):
    print(" ".join(
        word for word, _ in subtree.leaves()
    ))
```

```python

from sklearn.feature_extraction.text import CountVectorizer
from sklearn.decomposition import LatentDirichletAllocation
import pandas as pd
print("Rishabh 23bai70136")

# Step 1: Prepare documents
docs = [
    "machine learning models learn patterns from data",
    "artificial intelligence uses neural networks and data",
    "deep learning improves computer vision systems",
    "football team won the match and scored goals",
    "cricket players practiced batting and bowling",
    "the tournament team prepared for the final match",
    "doctor diagnosed the patient in the hospital",
    "medical treatment and medicine improved patient health",
    "hospital doctors analyze disease and treatment",
    "data science uses statistics and machine learning",
    "the football player trained for the tournament",
    "medical imaging helps doctors detect disease"
]

# Step 2: Convert text into word-count vectors
vectorizer = CountVectorizer(stop_words="english")
X = vectorizer.fit_transform(docs)

print("Document-Term Matrix Shape:", X.shape)

# Step 3: Create the LDA model
lda = LatentDirichletAllocation(
    n_components=3,
    random_state=42,
    learning_method="batch"
)

# Step 4: Train model and get document-topic proportions
doc_topic = lda.fit_transform(X)

# Step 5: Display top words for each topic
terms = vectorizer.get_feature_names_out()

print("\nTOPIC WORDS")
for topic_id, weights in enumerate(lda.components_):
    top_idx = weights.argsort()[-6:][::-1]
    print(f"Topic {topic_id}:",
          ", ".join(terms[i] for i in top_idx))

# Step 6: Display topic proportions
result = pd.DataFrame(
    doc_topic,
    columns=["Topic_0", "Topic_1", "Topic_2"]
)

print("\nDOCUMENT-TOPIC DISTRIBUTION")
print(result.round(3))
```

```python
# %pip install sklearn-crfsuite
import sklearn_crfsuite

# AIM: Named Entity Recognition using CRF

# Step 1: Training dataset
print("Rishabh 23bai70136")

train_sents = [
    [
        ("Ravi", "B-PER"), ("works", "O"), ("at", "O"),
        ("Infosys", "B-ORG"), ("in", "O"), ("Pune", "B-LOC")
    ],
    [
        ("Anita", "B-PER"), ("joined", "O"),
        ("Google", "B-ORG"), ("in", "O"), ("Delhi", "B-LOC")
    ],
    [
        ("Microsoft", "B-ORG"), ("opened", "O"),
        ("an", "O"), ("office", "O"), ("in", "O"),
        ("Hyderabad", "B-LOC")
    ]
]

# Step 2: Extract word features
def word_features(sent, i):
    word = sent[i][0]

    features = {
        "bias": 1.0,
        "word.lower": word.lower(),
        "word[-3:]": word[-3:],
        "is_title": word.istitle(),
        "is_upper": word.isupper(),
        "is_digit": word.isdigit()
    }

    if i > 0:
        features["prev.lower"] = sent[i - 1][0].lower()
    else:
        features["BOS"] = True

    if i < len(sent) - 1:
        features["next.lower"] = sent[i + 1][0].lower()
    else:
        features["EOS"] = True

    return features


# Step 3: Prepare training data
X_train = [
    [word_features(sentence, i)
     for i in range(len(sentence))]
    for sentence in train_sents
]

y_train = [
    [label for _, label in sentence]
    for sentence in train_sents
]

# Step 4: Initialize CRF model
crf = sklearn_crfsuite.CRF(
    algorithm="lbfgs",
    c1=0.1,
    c2=0.1,
    max_iterations=100
)

# Step 5: Train model
crf.fit(X_train, y_train)

# Step 6: Predict entities in the first sentence
test = [
    ("Ravi", "O"),
    ("visited", "O"),
    ("Delhi", "O")
]

X_test = [[
    word_features(test, i)
    for i in range(len(test))
]]

print("CRF labels:", crf.predict(X_test)[0])

# Step 7: Predict entities in the second sentence
test2 = [
    ("Meera", "O"), ("works", "O"), ("at", "O"),
    ("Wipro", "O"), ("in", "O"), ("Mumbai", "O")
]

X2 = [[
    word_features(test2, i)
    for i in range(len(test2))
]]

print("CRF labels 2:", crf.predict(X2)[0])
```
